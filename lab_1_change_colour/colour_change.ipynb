```python
#Функция, которая принимает диапозон (r: 0-255, g: 0-255, b: 0-255) и в этом диапозоне меняет цвет на задаваемый
def change_colour(file_name: str ,colours_range: list[list[int]], colour : list[int] | tuple[int, int, int], output_name: str | None = None) -> None:

    image = Image.open(file_name).convert("RGB")
    pixels = image.load()
    width, height = image.size

    new_colour = tuple(colour)

    (r_min, r_max), (g_min, g_max), (b_min, b_max) = colours_range

    for x in range(width):
        for y in range(height):
            r, g, b = pixels[x,y]
            if r_min <= r <= r_max and g_min <= g <= g_max and b_min <= b <= b_max:
                pixels[x, y] = new_colour
    
    save_path = output_name if output_name else file_name
    image.save(save_path)
```

```python
# Заменим оттенки красного (r: 100-255, g: 0-50, b: 0-50) на зеленый (0, 255, 0)
ranges = [
    [100, 255], # r
    [0, 50],    # g
    [0, 50]     # b
]
target_color = [0, 255, 0]

change_colour("input_image.jpg", ranges, target_color, "output.png")
```
